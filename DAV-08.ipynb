#Practical 8 :  Create interactive visualization using Plotly Express and Ploty Graph Objects

***Step  1 - import  Required libraries ***

```python
# import pandas
#import plotly Express
# import Plotly Graph Objects

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

** Step 2: Create a sommon dataset **

```python
# creating a dataset
data = {
    "Month" : ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    "Product" : ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    "Region" : ["North",'South','East','West',"North",'South','East','West',"North",'South','East','West'],
    "Sales" : [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold" : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    "Profit" : [18000,27000,30000,22000,32000,35000,28000,34000,38000,33000,46000,50000]
}
df = pd.DataFrame(data)
df
```

```python
#Display first five rows .
df.head()
```

```python
# dispaly the no of rows and column
df.shape
```

```python
# Display thr column names and datatypes
df.info()
```

**Step 3 create an interactive line chart using ploty Express**

```python
# How dose sales revenue change throughout the year?
fig = px.line(df, x="Month", y="Sales", title="Sales Revenue Over Time")
fig.show()
```

**step 4 : Improve the hover Information **

```python
# create  a line chart using sales over Month.
# Add Additional columns to the hover information . Students can see products ,Region , Unit Sold and profit when they move the mouse over the data

fig = px.line(df, x="Month", y="Sales", title="Sales Revenue Over Time",
             hover_name="Month", hover_data=["Units_Sold", "Profit"])
fig.show()
```

***Step 5 : Create an interactive Bar chart***

```python
 # Gourp the Dataset according to product.
df_grouped = df.groupby('Product').agg({'Sales': 'sum', 'Units_Sold': 'sum', 'Profit': 'sum'}).reset_index()
df_grouped
```

```python
 # calculate the total sales generated by each product
total_sales = df_grouped['Sales'].sum()
total_sales
```

```python
 # Create an interractive bar chart
 fig = px.bar(df_grouped, x='Product', y='Sales', title='Total Sales by Product')
 fig.show()
```

**Step 6 - interactive Scatter plot **

```python
# TO analyze the relationship between units sold and sales using an interractive scatter plot.
fig = px.scatter(df, x='Units_Sold', y='Sales', title='Units Sold vs Sales')
fig.show()
```

**Step 7 : Interactive Scatter plot with color **

```python
# create an scatter plot with color .
fig = px.scatter(df, x='Units_Sold', y='Sales', color='Product',title='Units Sold vs Sales')
fig.show()
```

Step 8 :  Create an interactive Pie Chart

```python
# To visualize regional sales contribution  using an interactive Pie Chart
fig = px.pie(df, values='Sales', names='Region', title='Sales by Region')
fig.show()
```

Step 9 :Plotly Graph Objects

```python
# ploting plotly graph

fig = go.Figure()

fig.add_trace(go.Line(
    x=df['Month'],
    y=df['Sales'],
    mode='lines+markers',
    name='Sales'
))

fig.show()
```
