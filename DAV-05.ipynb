Step 9 - Detect Outliers Using IQR( Select the Final_Score columns for outlier analysis.)

```python
# Select Final_Score column
final_score = df["Final_Score"]

# Calculate Q1 and Q3
Q1 = final_score.quantile(0.25)
Q3 = final_score.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

# Calculate outlier limits
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

```

Step 10 - Identify Outliers

1.   Select records where final_Score are below lower bound or above the upper bound
2.   Display the detected outliers records.



```python
# Identify outliers
outliers = df[
    (df["Final_Score"] < lower_bound) |
    (df["Final_Score"] > upper_bound)
]

# Display outliers
print("Outliers in Final_Score:")
print(outliers)
```

Step 11 - Count Outliers [ Count the total number of detected outliers.]

```python
# Count total number of outliers
outlier_count = len(outliers)

print("Total number of outliers:", outlier_count)
```
