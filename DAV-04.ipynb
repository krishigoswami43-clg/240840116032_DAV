Step: 1 Import Libraries

```python
# Import the required libraries
import pandas as pd

import matplotlib.pyplot as plt
```

Step:2 Create a Dataset

```python
#Create a dictionary containing monthly sales data such as Month, Units Sold, Revenue, and Profit.
data = {
    'Month': ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],
    'Units_Sold': [120, 135, 150, 145, 160, 175, 180, 170, 190, 210, 230, 250],
    'Revenue': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    'Profit': [20000, 23000, 25000, 24000, 27000, 30000, 31000, 29000, 33000, 37000, 41000, 45000]
}

df = pd.DataFrame(data)
df
```

step: 3 Inspect the dataset

```python
# Display the first five rows of the dataset
df.head()
```

```python
# Display the number of rows and columns
df.shape
```

```python
# Display the names of all columns
df.columns
```

```python
# Display detailed information
df.info()
```

Step: 4 Generate discriptive statistic

```python
# Calculate statistical measures
df.describe()
```

step 5 calculate mean

```python
# Calculate the average number of units sold during the year
average_units = df['Units_Sold'].mean()
print("Average Units Sold:", average_units)
```

Step: 6 calculate median

```python
# Find the middle value of the Units_Sold data
median_units = df['Units_Sold'].median()
print("Median Units Sold:", median_units)
```

Step: 7 calculate max and min

```python
# Find the highest and lowest number of units sold
maximum_units = df['Units_Sold'].max()
minimum_units = df['Units_Sold'].min()
print("Maximum Units Sold:", maximum_units)
print("Minimum Units Sold:", minimum_units)
```

Step: 8 calculate standard deviation

```python
# Calculate the standard deviation of Units_Sold to measure the variation in monthly sales.
std_units = df['Units_Sold'].std()
print("Standard Deviation of Units Sold:", std_units)
```

Step: 9 find the month with highest sales

```python
# find the index of the row having the maximum revenue
highest_sales_index = df['Revenue'].idxmax()
highest_sales_month = df.loc[highest_sales_index, 'Month']
highest_revenue = df.loc[highest_sales_index, 'Revenue']
print("Month with Highest Sales:", highest_sales_month)
print("Highest Revenue:", highest_revenue)
```

Step: 10 find the month with lowest sales

```python
# Find the index of the row having the minimum revenue
lowest_sales_index = df['Revenue'].idxmin()
lowest_sales_month = df.loc[lowest_sales_index, 'Month']
lowest_revenue = df.loc[lowest_sales_index, 'Revenue']
print("Month with Lowest Sales:", lowest_sales_month)
print("Lowest Revenue:", lowest_revenue)
```

Step:11 cal total reven

```python
# Calculate the total revenue generated during all 12 months.
total_revenue = df['Revenue'].sum()
print("Total Revenue:", total_revenue)
```

Step:12 cal total profit

```python
# Calculate the total profit earned during all 12 months.
total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
```

Step:13 identify the sales

```python
# Create a line chart to visualize the monthly revenue
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(df['Month'], df['Revenue'], marker='o')

plt.xlabel('Month')
plt.ylabel('Revenue')
plt.title('Monthly Revenue')
plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

Step:14 Analyze profit

```python
#create a line chart for monthly profit


import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(df['Month'], df['Profit'], marker='o')

plt.xlabel('Month')
plt.ylabel('Profit')
plt.title('Monthly Profit')

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

Step:15 compare revenue and profit both

```python
# Create a line chart containing both Revenue and Profit
plt.figure(figsize=(10, 5))

plt.plot(df['Month'], df['Revenue'], marker='o', label='Revenue')

plt.plot(df['Month'], df['Profit'], marker='o', label='Profit')

plt.xlabel('Month')
plt.ylabel('Amount')
plt.title('Monthly Revenue and Profit')

plt.xticks(rotation=45)

plt.legend()
plt.grid(True)

plt.show()
```

Step:16 Identify growth in sales

```python
# Calculate the percentage change in revenue from one month to the next month
df['Revenue_Growth_%'] = df['Revenue'].pct_change() * 100
df
```

Step:17 Identify the months with increasing revenue

```python
# Filter the dataset to find the months where revenue growth is greater than 0%
increasing_sales = df[df['Revenue_Growth_%'] > 0]
df
```
