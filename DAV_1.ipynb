Step 1: import the pandas library

```python
#Import the pandas library
import pandas as pd
```

Step 2: create sample **data**

```python
#Create sample data
data = {
    "ID" : [101,102,103,104,105,106,107],
    "Name" : ["Rahul","Priya","Amit","Riya","Karan","Krishna", "Harry"],
    "Age" : [20,21,19,20,22,24,25],
    "City" : [ "surat","navsari","valsad","vapi","bhruch","rajkot","surat"],
    "marks": [85,86,84,82,83,84,86]
}
```

Step 3: create a dataframe

```python
# Step 3: Create a DataFrame

df = pd.DataFrame(data)
df
```

Step 4: save data as a csv


```python
# Step 3: Create a DataFrame
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

Step 5: save data as an excel file

```python
# Step 5: Save the DataFrame as an Excel file
df.to_excel("students.xlsx", index=False)
print("Excle file created successfully.")
```

Step 6: save the data as a JSON file

```python
# Step 6: Save the DataFrame as a JSON file
df.to_json("students.json" , orient="records", indent=4)
print("JSON file created successfully.")
```
