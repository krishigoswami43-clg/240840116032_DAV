Aim : Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Task - 01 Create a Dataset

```python
import pandas as pd

data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Product": ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet",],
    "Region": ["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold": [12,30,25,14,35,30,18,38,32,21,45,40]
}

df = pd.DataFrame(data)

print(df)
```

Task - 02 Create  a line chart

- How did the company's sales change from Jan to Dec ?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    df["Month"],
    df["Sales"],
    marker="o",
    color="blue",
    linewidth=2
)

# Add title and labels
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

# Add grid
plt.grid(True, linestyle="--", alpha=0.6)

# Display chart
plt.show()
```

Task 02 Interpretation/Observation -
 The Line graph shows the Continuous Increase in the sales value From Jan to Dec

Task 03 - Create a Bar Chart

- Which product generated how much total sales?

```python
import matplotlib.pyplot as plt

# Bar chart for Product vs Sales
plt.figure(figsize=(8, 5))

plt.bar(df["Product"], df["Sales"], color="skyblue")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

Task 03 Interpretation/Observation -
- From Bar Chart we can see that the product Tablet has sold more or highest in comparision with other product

Task 04 - Create a Histogram

- How are the individual sales values distributed?

```python
import matplotlib.pyplot as plt

plt.hist(df["Sales"], bins=5, edgecolor="black")

plt.title("Histogram of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```


Task 04 Interpretation/Observation -
- From Histogram we can conclude that :Sales are mostly concentrated in the higher range, with sales increasing from around ₹1,20,000 to ₹2,50,000. This indicates an overall upward sales pattern in the dataset.

Task 05 - Create a Scatter Plot

- Is there a relationship between the number of units sold and total sales?

```python
import matplotlib.pyplot as plt

plt.scatter(df["Units_Sold"], df["Sales"])

plt.title("Scatter Plot of Units Sold vs Sales")
plt.xlabel("Units Sold")
plt.ylabel("Sales")

plt.show()
```

Task 05 Interpretation/Observation -
- The scatter plot shows a positive relationship between Units Sold and Sales. As the number of units sold increases, sales also generally increase. This suggests that selling more units is associated with higher total sales.
- However, the points are somewhat scattered, so the relationship is not perfectly strong or linear.

Task 06 - Create a Pie Chart

- What percentage of total sales comes from each region?

```python
import matplotlib.pyplot as plt

# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum()

# Create pie chart
plt.pie(
    region_sales,
    labels=region_sales.index,
    autopct="%1.1f%%"
)

plt.title("Percentage of Total Sales by Region")
plt.show()
```

Task 06 Interpretation/Observation -
- From the Pie Chart of Total Sales by Region we can conclude that : all regions are contributing near to the same east and west are just a percentage more and north is just a little percentage less .
- Overall, there is not a very large difference in sales contribution among the four regions.
